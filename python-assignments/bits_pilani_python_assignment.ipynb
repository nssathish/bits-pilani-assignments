```python
"""Given a number, find the sum of its digits. Take the number as an input from the user."""
num = int(input("Enter a number: "))

def sum_of_digits(_number: int) -> int:
    """ Calculates the sum of digits of an input number """
    _sum = 0
    while _number > 0:
        _sum += _number % 10
        _number //= 10

    return _sum

print(sum_of_digits(num))
```

```python
#
# Given a number, check whether the given number is an Armstrong number or not.
# A positive integer is called an Armstrong number of order n if:
# abcd... = an + bn + cn + dn + ...
# Example: 153 = 1*1*1 + 5*5*5 + 3*3*3
# 153 is an Armstrong number of order 3.
# Inputs from the user will be number and order n.
#

num = int(input("Enter a number: "))
order = int(input("Enter order: "))

def is_armstrong(_number: int, n: int) -> bool:
    """check whether the given number is an armstrong number"""

    _sum = 0
    _num = _number
    while _number > 0:
        _sum += ((_number % 10) ** n)
        _number //= 10

    return _sum == _num

print(is_armstrong(num, order))
```

```python
# Given a string, write a python function to check if it is a palindrome or not.
# A string is said to be palindrome if the reverse of the string is the same as string.
# For example, “malayalam” is a palindrome, but “music” is not a palindrome.

_string = input("Enter a string: ")

def is_palindrome(string: str) -> bool:
    """ check if the given string is a palindrome """
    start = 0
    end = len(string)

    while start < end:
        if string[start] != string[end]:
            return False
        start += 1
        end -= 1

    return True

print(is_palindrome(_string))
```

```python
# Input a text file (containing 1 or more paragraphs of English text) from the user,
# parse this file to display the frequency of occurrence of each word in this text file.
# Find the 3 most frequent words as well.
from pprint import pprint
from itertools import islice

file_loc = input("Enter file location: ")

def freq_of_occurrence(_file_loc: str) -> dict:
    freq_ledger = {}

    with open(_file_loc, "r") as _f:
        for line in _f:
            _words: list[str] = line.strip().lower().split(" ")
            for word in _words:
                if word not in freq_ledger:
                    freq_ledger[word] = 1
                else:
                    freq_ledger[word] += 1

        freq_ledger = dict(sorted(freq_ledger.items(), key=lambda item:item[1], reverse=True))
    
    return freq_ledger

_frequencies = freq_of_occurrence(file_loc)
pprint(_frequencies, sort_dicts=False)
print(f"Top 3 most frequent words: {list(islice(_frequencies.keys(), 3))}")
```
